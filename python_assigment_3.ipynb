```python
import matplotlib.pyplot as plt

completed_prep = df[df['test preparation course'] == 'completed']['math score']
none_prep = df[df['test preparation course'] == 'none']['math score']

fig, axes = plt.subplots(1, 2, figsize=(12, 6), sharey=True)

axes[0].hist(completed_prep, bins=20, edgecolor='black')
axes[0].set_title('Math Scores - Completed Test Prep')
axes[0].set_xlabel('Math Score')
axes[0].set_ylabel('Frequency')

axes[1].hist(none_prep, bins=20, edgecolor='black')
axes[1].set_title('Math Scores - No Test Prep')
axes[1].set_xlabel('Math Score')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

ethnic_group_scores = df.groupby('race/ethnicity')[['math score', 'reading score', 'writing score']].mean()

ethnic_group_scores.T.plot(kind='line', figsize=(10, 6))
plt.xlabel('Subjects')
plt.ylabel('Average Score')
plt.title('Average Scores by Ethnic Group')
plt.show()
```

```python
import matplotlib.pyplot as plt

gender_scores = df.groupby('gender')[['math score', 'reading score', 'writing score']].mean()
print(gender_scores)
gender_scores.plot(kind='bar', stacked=True, figsize=(10, 6))
plt.xlabel('Gender')
plt.ylabel('Average Score')
plt.title('Average Scores by Gender (Stacked)')
plt.xticks(rotation=0)
plt.show()
```

```python
import matplotlib.pyplot as plt

lunch_counts = df['lunch'].value_counts()
print(lunch_counts)
plt.figure(figsize=(8, 8))
plt.pie(lunch_counts, labels=lunch_counts.index,  startangle=90, colors=['blue', 'red'])
plt.title('Proportion of Students by Lunch Type')
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

axes[0].hist(df['math score'], bins=20, edgecolor='black')
axes[0].set_title('Math Scores')
axes[0].set_xlabel('Score')
axes[0].set_ylabel('Frequency')

axes[1].hist(df['reading score'], bins=20, edgecolor='black')
axes[1].set_title('Reading Scores')
axes[1].set_xlabel('Score')
axes[1].set_ylabel('Frequency')

axes[2].hist(df['writing score'], bins=20, edgecolor='black')
axes[2].set_title('Writing Scores')
axes[2].set_xlabel('Score')
axes[2].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
